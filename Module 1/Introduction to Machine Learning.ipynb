```python
asian_cars = df[df['origin'] == 'Asia']
asian_cars = asian_cars[['vehicle_weight', 'model_year']]
asian_cars = asian_cars.head(7)
X = asian_cars.values
```

```python
XTX = X.T @ X
XTX = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
W = XTX @ X.T @ y
np.sum(W)
```
